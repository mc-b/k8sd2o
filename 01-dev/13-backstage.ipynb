## Backstage

[Backstage.io](https://backstage.io) ist eine Open-Source-Developer-Plattform, die von Spotify entwickelt wurde, um die Verwaltung von Softwareprojekten zu vereinfachen. 

Mit Backstage können Unternehmen eine zentrale Anlaufstelle für all ihre internen Tools, Services, Dokumentationen und Software-Komponenten schaffen. 

Im Mittelpunkt steht das Konzept des "Service Catalogs", der alle Applikationen und Services übersichtlich darstellt. 

Durch Plugins lässt sich Backstage flexibel erweitern und an individuelle Bedürfnisse anpassen. Ziel ist es, Entwickler:innen die tägliche Arbeit zu erleichtern und eine einheitliche Nutzererfahrung über verschiedene Tools hinweg zu bieten.

Für einen kleinen Überblick verwenden wir die vorbereitete Backstage Umgebung vom [AutoShop](https://gitlab.com/ch-mc-b/autoshop-ms/infra/backstage/-/tree/main/1.51)


```python
%%bash
kubectl apply -f - <<EOF
apiVersion: v1
kind: Pod
metadata:
  name: backstage
  labels:
    app: backstage
spec:
  containers:
    - name: backstage
      image: registry.gitlab.com/ch-mc-b/autoshop-ms/infra/backstage/backstage:1.51
      imagePullPolicy: Always
      ports:
        - containerPort: 7007
      env:
        - name: BACKSTAGE_ORG
          value: "AutoShop K8s"
        - name: BACKSTAGE_PORT
          value: "7007"
        - name: BACKSTAGE_HOST
          value: "$(cat ~/data/server-ip)"                    
---
apiVersion: v1
kind: Service
metadata:
  name: backstage
spec:
  type: NodePort
  selector:
    app: backstage
  ports:
    - name: http
      port: 7007
      targetPort: 7007
      nodePort: 30007
EOF
```

Über folgenden URL können wie auf die Oberfläche zugreifen

```python
%%bash
echo "URL: http://$(cat ~/data/server-ip):30007"
```

---

### Aufräumen

```python
%%bash
kubectl delete pod/backstage
kubectl delete service/backstage
```
