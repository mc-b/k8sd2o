## Gitlab

Installation

```python
%%bash
curl https://packages.gitlab.com/install/repositories/gitlab/gitlab-ce/script.deb.sh | sudo os=ubuntu dist=noble bash - 
sudo EXTERNAL_URL="http://$(cat /home/ubuntu/data/server-ip):9999" apt-get install -y gitlab-ce
```

### Gitlab CE

Die Gitlab CE (Community Edition) kann unter dem folgenden URL errreicht werden:

```python
%%bash
echo "URL     : http://$(cat ~/data/server-ip):9999"
echo "User    : root"
echo "$(sudo cat /etc/gitlab/initial_root_password  | grep Password:)"
```
