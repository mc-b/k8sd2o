## Controller – Grundprinzip

(Quelle: Kapitel 27 )

**Definition:**
Ein Controller ist ein Prozess, der kontinuierlich den *Ist-Zustand* von Kubernetes-Ressourcen mit dem *Soll-Zustand* vergleicht und Abweichungen automatisch behebt (Reconciliation Loop).

**Zentraler Ablauf (Observe – Analyze – Act):**

1. **Observe**: Ereignisse (Events) überwachen, z. B. wenn sich eine Ressource ändert.
2. **Analyze**: Prüfen, ob der Zustand vom Zielzustand abweicht.
3. **Act**: Aktionen ausführen (z. B. Pods starten, löschen, aktualisieren).

**Beispiele für eingebaute Controller:**

* Deployment Controller
* ReplicaSet Controller
* StatefulSet Controller
* DaemonSet Controller

**Custom Controller:**

* Reagieren auf Änderungen in Standardressourcen wie ConfigMaps, Secrets, Pods etc.
* Können Logik zur Laufzeit ergänzen, ohne Kubernetes selbst verändern zu müssen.
* Lassen sich in jeder Sprache umsetzen (Go gängig, aber z. B. auch Shell-Skripte möglich).

**Beispiel im PDF:**
Ein kleines Shell-basiertes Skript beobachtet Änderungen an ConfigMaps und startet betroffene Pods neu, wenn eine bestimmte Annotation gesetzt ist.

**Wichtig:**
Custom Controller arbeiten *nur* mit bestehenden Kubernetes-Ressourcen – **keine neuen API-Typen**.
