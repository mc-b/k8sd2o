## Operator – Erweiterung des Controller-Konzepts

(Quelle: Kapitel 28 )

**Definition:**
Ein Operator ist ein spezialisierter Controller, der zusätzlich **Custom Resource Definitions (CRDs)** verwendet, um anwendungs- oder domänenspezifische Logik in Kubernetes abzubilden.

**Wesentliche Merkmale:**

* Fügt Kubernetes **neue Ressourcentypen** hinzu (z. B. Prometheus, KafkaCluster, BackupJob).
* Kapselt Expertenwissen (Operational Intelligence) einer Anwendung in Code.
* Ermöglicht vollständiges Lifecycle-Management:

  * Installation
  * Konfiguration
  * Updates
  * Backups
  * Healing / Failover

**CRD-Struktur:**
Ein CRD definiert:

* API-Gruppe
* Versionen
* Scope (Namespaced/Cluster)
* Schema (OpenAPI v3)
* Optional: Subresources wie `/status` oder `/scale`

**Typen von CRDs:**

* **Installation CRDs** (z. B. zur Installation von Prometheus selbst)
* **Application CRDs** (z. B. ServiceMonitor für Prometheus)

**Beispiel im PDF:**
Ein *ConfigWatcher* Operator ersetzt den vorherigen Annotation-basierten Controller:

* CRD beschreibt, welche ConfigMaps zu beobachten sind
* Operator liest CRD und führt passende Aktionen (z. B. Pod-Restarts) aus

Damit löst man Einschränkungen des reinen Controllers (z. B. keine Mehrfachzuweisung, keine Validierung).

**Frameworks für Operator-Entwicklung:**

* **Kubebuilder** (Go, Standard)
* **Operator SDK** (inkl. Go, Ansible, Helm, Java)
* **Metacontroller** (Webhook-basierte Logik, sprachunabhängig)
